```python
def_remoto = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Remoto'
}

df['remoto'] = df['remoto'].replace(def_remoto)
df['remoto'].value_counts()
```

## Criando DFs para entender novos conceitos

```python
import numpy as np

#Criando um novo DF para entender o fillna -> media -> mediana

df_salarios = pd.DataFrame({

    'nome':["ana", "bruno", "carla", "daniel", "edson"],
    'salario' : [4000, np.nan, 5000, np.nan, 100000]

})

df_salarios['salario_media'] = df_salarios['salario'].fillna(df_salarios['salario'].mean().round(2))
df_salarios['salario_mediana'] = df_salarios['salario'].fillna(df_salarios['salario'].median())

df_salarios
```

```python
#Novo DF pra entender o ffil (forward fill = completa null com o valor anterior)

df_temperaturas = pd.DataFrame({

    'dia':['seg','ter','qua', 'qui', 'sex'],
    'temperatura':[30, np.nan, np.nan, 28, 27]

})

df_temperaturas['preenchido_ffill'] = df_temperaturas['temperatura'].ffill()
df_temperaturas
```

```python
#Novo DF pra entender o bfil (backward fill = completa null com o valor posterior)

df_temperaturas = pd.DataFrame({

    'dia':['seg','ter','qua', 'qui', 'sex'],
    'temperatura':[30, np.nan, np.nan, 28, 27]

})

df_temperaturas['preenchido_bfill'] = df_temperaturas['temperatura'].bfill()
df_temperaturas
```

```python
#Novo DF pra entender fillna com valor específico

df_cidades = pd.DataFrame({

    'nome':["ana", "bruno", "carla", "daniel", "edson"],
    'cidade':['sao paulo', np.nan, 'curitiba', np.nan, 'belem']
    
})

df_cidades['cidade_preenchida'] = df_cidades['cidade'].fillna('não informado')
df_cidades
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data = df_limpo, x='senioridade', y='usd')
plt.title('Salário anual por nível de senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Salário')
plt.show()
```

```python
#Calculo da média do salário pra cada senioridade do maior pro menor
df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False)
```

```python
ordem = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=True).index
ordem
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data = df_limpo, x='senioridade', y='usd', order=ordem)
plt.title('Salário anual por nível de senioridade')
plt.xlabel('Senioridade')
plt.ylabel('Salário')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(df_limpo['usd'],bins=50, kde=True)
plt.title('Distribuição dos salários anuais')
plt.xlabel('Salário em USD')
plt.ylabel('Frequência')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df_limpo['usd'])
plt.title('Boxplot salário')
plt.xlabel('Salário em USD')
plt.show()
```

```python
ordem_senioridade = ['Júnior','Pleno','Sênior','Executivo']
plt.figure(figsize=(8,8))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade)
plt.title('Distribuição salarial por senioridade')
plt.xlabel('Senioridade')
plt.show()
```

```python
ordem_senioridade = ['Júnior','Pleno','Sênior','Executivo']
plt.figure(figsize=(8,8))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade, palette='Set2', hue='senioridade')
plt.title('Distribuição salarial por senioridade')
plt.xlabel('Senioridade')
plt.show()
```

```python
import plotly.express as px
```

```python
senioridade_media_salario = df_limpo.groupby('senioridade')['usd'].mean().sort_values(ascending=False).reset_index()
```

```python
fig = px.bar(senioridade_media_salario, 
             x='senioridade', 
             y='usd',
             title='Média Salarial por Senioridade',
             labels={'usd': 'Salário anual (USD)', 'senioridade': 'Senioridade'},
             category_orders={'senioridade': ordem_senioridade})
fig.show()
```

```python
remoto_contagem = df_limpo['remoto'].value_counts().reset_index()
remoto_contagem.columns = ['tipo_trabalho', 'quantidade']
remoto_contagem
```

```python
fig = px.pie(remoto_contagem, 
             names='tipo_trabalho', 
             values='quantidade',
             title='Proporção de modalidades de trabalho'
             )
fig.show()
```

```python
fig = px.pie(remoto_contagem, 
             names='tipo_trabalho', 
             values='quantidade',
             title='Proporção de modalidades de trabalho',
             hole=0.5
             )
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show()
```
